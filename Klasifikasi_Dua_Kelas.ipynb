```python
# Model untuk klasifikasi dua kelas
# Oranges vs. Grapefruit
# https://www.kaggle.com/datasets/joshmcadams/oranges-vs-grapefruit
```

```python
# Menggunakan dataset lokal
import pandas as pd
# df = pd.read_csv('citrus.csv')

# Menggunakan dataset dari google drive
df = pd.read_csv('https://drive.google.com/uc?id=1XZbZ5Z7MBXYaR3UO1HbQ4ncPi4R3HsyF')
df.head()
```

```python
df.info()
```

```python
# Mengubah column 'name' menjadi numerik
# 0 = grapefruit ; 1 = orange

# Cara pertama (untuk kasus sederhana)
# df.name[df.name == 'grapefruit'] = 0
# df.name[df.name == 'orange'] = 1

# Cara kedua (jika labelnya ada banyak)
from sklearn import preprocessing

label_encoder = preprocessing.LabelEncoder()
df['name'] = label_encoder.fit_transform(df['name'])
print(df)
```

```python
df.info()
```

```python
# Dataset harus dalam bentuk arrat agar dapat diproses oleh model
# Attribute values mengembalikan numpy array yang dikonversi dari dataframe
dataset = df.values
dataset
```

```python
"""
pilih 5 kolom terakhir sebagai atribut
bilangan sebelum koma untuk memilih baris pada dataframe
bilangan setelah koma untuk memilih kolom pada dataframe
"""
X = dataset[:,1:6]
```

```python
Y = dataset[:,0]
```

```python
# Normalisasi data, agar rentangnya 0-1
from sklearn import preprocessing
min_max_scaler = preprocessing.MinMaxScaler()
X_scale = min_max_scaler.fit_transform(X)
X_scale
```

```python
# Split data training dan testing
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X_scale, Y, test_size = 0.3)
```

```python
# MASUK KE TAHAP MEMBANGUN MODEL NEURAL NETWORK
from keras.models import Sequential
from keras.layers import Dense
```

```python
# Model neural network
model = Sequential([
    Dense(32, activation='relu', input_shape=(5,)),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid'),
])
```

```python
# Menentukan optimizer dan loss function
model.compile(optimizer = 'sgd',
              loss = 'binary_crossentropy',
              metrics = ['accuracy'])
```

```python
# Pelatihan model
model.fit(X_train, Y_train, epochs = 100)
```

```python
# Melihat performa model pada data baru
model.evaluate(X_test, Y_test, batch_size = 1)
```
